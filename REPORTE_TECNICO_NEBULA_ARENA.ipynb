### 6. Recomendaciones Técnicas de Producción
1. **Precalentamiento de Flota (Warm Buffer):** Programar escalamiento preventivo 15 min antes de torneos clave.
2. **Empaquetado Packed:** Máxima densidad (6 servidores de 10 jugadores por nodo `e2-standard-4`).
3. **Origin Shielding:** Obligatorio en Cloud CDN para soportar hasta 20,000 espectadores con solo 135 Mbps de tráfico a origen.
4. **CUDs (Committed Use Discounts):** Ahorro del 57% en la base de cómputo permanente.
